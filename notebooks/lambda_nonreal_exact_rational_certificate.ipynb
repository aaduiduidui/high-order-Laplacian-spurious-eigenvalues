# Exact rational certificate for condition (d)

This notebook verifies condition (d) in Theorem 5, namely

$$
    \lambda^\ast\notin\mathbb R.
$$

After eliminating the radical, the candidate eigenvalue is represented as a rational function of $s_0^\ast$:

$$
    \lambda^\ast=\frac{f_6(s_0^\ast)}{f_7(s_0^\ast)}.
$$

All arithmetic in this notebook is exact. No floating-point arithmetic is used.


## Target certificate

Let

$$
    F_6=f_6(\widehat{s}_0), \qquad F_7=f_7(\widehat{s}_0),
$$

where $\widehat{s}_0$ is the rational corner point of the isolating rectangle $I$. We prove:

$$
    f_7(s_0^\ast)\neq 0,
    \qquad
    \operatorname{Im}\frac{f_6(s_0^\ast)}{f_7(s_0^\ast)}>0.
$$

The second inequality implies $\lambda^\ast\notin\mathbb R$.

If

$$
    f_6(s_0^\ast)=A'+iB', \qquad f_7(s_0^\ast)=C'+iD',
$$

then

$$
    \operatorname{Im}\frac{f_6(s_0^\ast)}{f_7(s_0^\ast)}
    =
    \frac{B'C'-A'D'}{|f_7(s_0^\ast)|^2}.
$$

Thus it suffices to verify $f_7(s_0^\ast)\neq0$ and

$$
    B'C'-A'D'>0.
$$


```python
from fractions import Fraction

from rational_tools import (
    im_div_numerator,
    polynomial_perturbation_bound,
    poly_eval,
    rectangle_distance_bound,
    rectangle_modulus_bound,
    require,
)
from shared_certificate_data import xL, xU, yL, yU, s_hat, delta, R, f6, f7
```

## Rectangle checks

The rectangle is

$$
    I=\{x+iy:x_L\le x\le x_U,\ y_L\le y\le y_U\}.
$$

The rational point

$$
    \widehat{s}_0=\frac{160734}{10^9}-\frac{6419165}{10^9}i
$$

is used as the expansion point. The certificate uses the rational overestimates

$$
    |s_0^\ast-\widehat{s}_0|\leq \delta=\frac{2}{10^9},
    \qquad
    |s_0^\ast|\leq R=\frac{1}{100}.
$$


```python
print("Step 1: specifying the rational rectangle I for s_0 = x + i y")
print("Thus I = {x + i y : xL <= x <= xU, yL <= y <= yU}.")
print("xL: lower bound for Re(s_0) =", xL)
print("xU: upper bound for Re(s_0) =", xU)
print("yL: lower bound for Im(s_0) =", yL)
print("yU: upper bound for Im(s_0) =", yU)
print("s_hat:", s_hat)
print("delta:", delta)
print("R:", R)

require(
    "the rectangle I lies within distance delta of s_hat",
    rectangle_distance_bound(xL, xU, yL, yU, s_hat, delta),
)
require(
    "the rectangle I lies in |z| <= R",
    rectangle_modulus_bound(xL, xU, yL, yU, R),
)
```

## Perturbation bounds for $f_6$ and $f_7$

For a polynomial $p(z)=\sum_k a_k z^k$, the common perturbation estimate gives

$$
    |p(s_0^\ast)-p(\widehat{s}_0)|
    \leq
    \delta\sum_{k\geq1}|a_k|kR^{k-1}.
$$

We apply this to $f_6$ and $f_7$.


```python
F6 = poly_eval(f6, s_hat)
F7 = poly_eval(f7, s_hat)
E6 = polynomial_perturbation_bound(f6, delta, R)
E7 = polynomial_perturbation_bound(f7, delta, R)

print("F6 = f6(s_hat):", F6)
print("F7 = f7(s_hat):", F7)
print("E6: perturbation bound for f6 on I =", E6)
print("E7: perturbation bound for f7 on I =", E7)
```

## Denominator nonvanishing

It is enough to verify

$$
    E_7^2<|F_7|^2.
$$

Then

$$
    |f_7(s_0^\ast)|\geq |F_7|-E_7>0.
$$


```python
require(
    "f7(s0*) is nonzero: E7^2 < |F7|^2",
    E7 * E7 < F7.norm2(),
)
```

## Sign of the imaginary part

Write

$$
    F_6=A+iB, \qquad F_7=C+iD.
$$

The exact numerator of $\operatorname{Im}(F_6/F_7)$ is

$$
    N_0=BC-AD.
$$

For the true values

$$
    f_6(s_0^\ast)=A'+iB', \qquad f_7(s_0^\ast)=C'+iD',
$$

we use the bound

$$
\begin{aligned}
 |(B'C'-A'D')-(BC-AD)|
&\le |B'-B||C|+|B||C'-C|+|B'-B||C'-C| \\
&\quad +|A'-A||D|+|A||D'-D|+|A'-A||D'-D| \\
&\le E_6|C|+|B|E_7+E_6E_7
    +E_6|D|+|A|E_7+E_6E_7 \\
&=(|A|+|B|)E_7+(|C|+|D|)E_6+2E_6E_7.
\end{aligned}
$$

We denote this upper bound by $E_N$.
If $N_0-E_N>0$, then $B'C'-A'D'>0$.


```python
A, B = F6.re, F6.im
C, D = F7.re, F7.im

N0 = im_div_numerator(F6, F7)
EN = (abs(A) + abs(B)) * E7 + (abs(C) + abs(D)) * E6 + 2 * E6 * E7

print("N0 = numerator of Im(F6/F7) at s_hat:", N0)
print("EN = perturbation bound for the numerator:", EN)

require(
    "the imaginary numerator remains positive on I: N0 - EN > 0",
    N0 - EN > 0,
)
```

## Conclusion

The denominator $f_7(s_0^\ast)$ is nonzero, and the numerator of the imaginary part of $f_6(s_0^\ast)/f_7(s_0^\ast)$ is positive. Therefore

$$
    \operatorname{Im}\lambda^\ast
    =
    \operatorname{Im}\frac{f_6(s_0^\ast)}{f_7(s_0^\ast)}
    >0.
$$

In particular,

$$
    \lambda^\ast\notin\mathbb R.
$$

This verifies condition (d) for the plus-sign candidate. The conjugate candidate is handled by complex conjugation.
